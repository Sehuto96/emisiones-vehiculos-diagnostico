## 2. Carga de datos
Se cargan los dos datasets del proyecto, cada uno en su propio DataFrame:
- `df_co2`: CO2 Emission by Vehicles (Kaggle, datos de Canadá)
- `df_epa`: Vehicle Fuel Economy (Data.gov / EPA, datos de EE. UU.)

```python
# Paso 1: importar Pandas, la librería para manejar tablas de datos
import pandas as pd

# Paso 2: cargar el Dataset 1 (Kaggle, Canadá) en el DataFrame df_co2
df_co2 = pd.read_csv("../data/raw/CO2 Emissions_Canada.csv")

# Paso 3: cargar el Dataset 2 (EPA, EE. UU.) en el DataFrame df_epa
# low_memory=False hace que Pandas lea todo el archivo antes de decidir
# el tipo de cada columna (el archivo es grande)
df_epa = pd.read_csv("../data/raw/fuel_economy_epa.csv", low_memory=False)

# Paso 4: verificar que ambos se cargaron (filas, columnas)
print("Dataset 1 (df_co2):", df_co2.shape)
print("Dataset 2 (df_epa):", df_epa.shape)
```
