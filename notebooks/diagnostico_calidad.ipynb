## 4. Diagnóstico de problemas de calidad
Antes de construir la tabla de problemas se verifican tres posibles hallazgos adicionales: vehículos con cero emisiones, valores atípicos en las emisiones y valores de relleno en `df_epa`.

```python
# Verificación 1: vehículos con 0 g/mi de CO2 por el escape, por tipo de combustible
# (los eléctricos y de hidrógeno no emiten por el escape y pueden sesgar los promedios)
df_epa[df_epa["co2TailpipeGpm"] == 0]["fuelType"].value_counts()
```

```python
# Verificación 2: valores atípicos en CO2 Emissions(g/km) de df_co2 con el método IQR
# Se consideran atípicos los valores por fuera de [Q1 - 1.5*IQR, Q3 + 1.5*IQR]
q1 = df_co2["CO2 Emissions(g/km)"].quantile(0.25)
q3 = df_co2["CO2 Emissions(g/km)"].quantile(0.75)
iqr = q3 - q1
lim_inf, lim_sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
atipicos_co2 = df_co2[(df_co2["CO2 Emissions(g/km)"] < lim_inf) |
                      (df_co2["CO2 Emissions(g/km)"] > lim_sup)]
print("Límite inferior:", lim_inf, "| Límite superior:", lim_sup)
print("Registros atípicos:", len(atipicos_co2),
      f"({len(atipicos_co2) / len(df_co2) * 100:.2f} % del dataset)")
```

```python
# Verificación 3: valores de relleno (-1) en la columna 'co2' de df_epa
# Un -1 no es un nulo de Pandas, por eso info() no lo detecta ("nulo escondido")
relleno = df_epa[df_epa["co2"] == -1]
print("Registros con co2 = -1:", len(relleno),
      f"({len(relleno) / len(df_epa) * 100:.2f} % del dataset)")
if len(relleno) > 0:
    print("Años afectados:", relleno["year"].min(), "a", relleno["year"].max())
```

```python
# Verificación 4: diferencia de unidades entre los dos datasets
# df_co2 mide emisiones en g/km; df_epa en g/mi (1 milla = 1.609 km)
print("df_co2 - CO2 Emissions(g/km), media:", round(df_co2["CO2 Emissions(g/km)"].mean(), 1))
print("df_epa - co2TailpipeGpm (g/mi), media:", round(df_epa["co2TailpipeGpm"].mean(), 1))
print("df_epa convertido a g/km, media:", round((df_epa["co2TailpipeGpm"] / 1.609).mean(), 1))
```
