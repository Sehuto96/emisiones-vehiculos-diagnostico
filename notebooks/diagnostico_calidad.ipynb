**Observaciones del Dataset 1 (df_co2):**
- Tiene 7385 filas y 12 columnas. Cada fila representa una configuración de vehículo (marca, modelo y características) con su consumo y sus emisiones de CO₂.
- Tipos de datos: 4 columnas float64, 3 int64 y 5 de texto (str/object). Variables numéricas (7): Engine Size(L), Cylinders, Fuel Consumption City (L/100 km), Fuel Consumption Hwy (L/100 km), Fuel Consumption Comb (L/100 km), Fuel Consumption Comb (mpg) y CO2 Emissions(g/km), esta última es la variable objetivo de la pregunta analítica. Los tipos son coherentes con lo que representan.
- Variables de texto (5): Make (42 marcas; la más frecuente es FORD con 628 registros), Model (2053 valores distintos), Vehicle Class (16 clases; la más común es SUV - SMALL con 1217), Transmission (27 códigos; el más frecuente es AS6 con 1324) y Fuel Type (5 códigos; X, gasolina regular, domina con 3637).
- Valores faltantes: las 12 columnas tienen 7385 valores no nulos, es decir, no hay valores faltantes.
- Rangos en describe(): el tamaño del motor va de 0.9 a 8.4 L (media 3.16); los cilindros de 3 a 16 (media 5.6); el consumo combinado de 4.1 a 26.1 L/100 km; y las emisiones de 96 a 522 g/km (media 250.6 y mediana 246). No aparecen ceros ni valores negativos, y los rangos son plausibles para vehículos desde compactos hasta deportivos de alto desempeño. El máximo de emisiones (522) queda lejos del percentil 75 (288), por lo que conviene revisar posibles valores atípicos más adelante.
- Duplicados: se encontraron 1103 filas duplicadas con duplicated().sum(), cerca del 14,9 % de los registros. Pueden distorsionar promedios y correlaciones, por lo que deberán revisarse antes del análisis.
- Nombres de columnas: tienen espacios, paréntesis, mayúsculas y símbolos (por ejemplo "Engine Size(L)" y "CO2 Emissions(g/km)"), con formatos poco uniformes. Convendrá estandarizarlos más adelante.
- Códigos: Transmission y Fuel Type usan abreviaturas (AS6, M6; X, Z, D, E, N) que solo se entienden con el diccionario de datos.
- Limitaciones: el consumo aparece en dos unidades (L/100 km y mpg), lo que duplica información y podría generar redundancia al analizar. Además, el dataset no incluye año del modelo ni vehículos eléctricos o híbridos (los tipos de combustible son gasolina, diésel, etanol y gas natural).
